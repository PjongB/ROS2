```python
from PIL import Image


def safe_call(name, func):
    try:
        func()
        print(f"[OK] {name}")
    except Exception as err:
        print(f"[WARN] {name} 실패: {err}")


def main():
    def led_off():
        from pinkylib import LED
        led = LED()
        if hasattr(led, "clear"):
            led.clear()
        else:
            led.fill((0, 0, 0))

    def lcd_off():
        from pinky_lcd.pinky_lcd import LCD
        lcd = LCD()
        if hasattr(lcd, "clear"):
            lcd.clear()
        else:
            black = Image.new("RGB", (320, 240), (0, 0, 0))
            lcd.img_show(black)

    def motor_stop():
        from pinkylib import Motor
        motor = Motor()
        if hasattr(motor, "enable_motor"):
            motor.enable_motor()
        motor.stop()

    safe_call("motor.stop()", motor_stop)
    safe_call("led off", led_off)
    safe_call("lcd off", lcd_off)

    print("완료: LED/LCD/Motor 종료 처리를 시도했습니다.")


if __name__ == "__main__":
    main()
```
