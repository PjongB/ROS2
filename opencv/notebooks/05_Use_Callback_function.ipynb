## 마우스 이벤트 확인

아래 예제는 데스크톱 OpenCV 창을 사용합니다. ESC를 누르면 종료됩니다.


```python
import cv2
import numpy as np

old_x = old_y = -1
image = np.ones((480, 640, 3), dtype=np.uint8) * 255

def on_mouse(event, x, y, flags, param):
    global old_x, old_y
    if event == cv2.EVENT_LBUTTONDOWN:
        old_x, old_y = x, y
        print(f"LBUTTON DOWN: {x}, {y}")
    elif event == cv2.EVENT_LBUTTONUP:
        print(f"LBUTTON UP: {x}, {y}")
    elif event == cv2.EVENT_MOUSEMOVE and flags & cv2.EVENT_FLAG_LBUTTON:
        cv2.line(param, (old_x, old_y), (x, y), (0, 0, 255), 4, cv2.LINE_AA)
        old_x, old_y = x, y

cv2.namedWindow("image")
cv2.setMouseCallback("image", on_mouse, image)

while True:
    cv2.imshow("image", image)
    if cv2.waitKey(20) & 0xFF == 27:
        break

cv2.destroyAllWindows()
```

## 콜백 핵심

- 콜백 함수 형식: `callback(event, x, y, flags, param)`
- `event`: 클릭·이동 같은 사건
- `flags`: 버튼이 눌린 상태
- `param`: 콜백에 전달할 사용자 데이터
