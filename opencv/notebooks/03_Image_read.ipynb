## 라이브러리 준비


```python
import matplotlib.pyplot as plt
import cv2
```

## 이미지 읽기


```python
img = cv2.imread('../data/lenna.bmp')
type(img)
```

## 잘못된 경로 확인


```python
tmp = cv2.imread('./dddd.jpg')
type(tmp)
```

## 영상 크기


```python
img.shape
```

## BGR을 RGB로 변환


```python
fix_img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(fix_img)
```

## 그레이 영상 읽기


```python
img_gray = cv2.imread('../data/lenna.bmp', cv2.IMREAD_GRAYSCALE)
plt.imshow(img_gray, cmap='gray')
```

## 컬러 영상 크기


```python
img.shape
```

## 크기 변경


```python
new_img = cv2.resize(fix_img, (800, 600))
plt.imshow(new_img)
```

## 상하 반전


```python
plt.imshow(cv2.flip(fix_img, 0))
```

## 좌우 반전


```python
plt.imshow(cv2.flip(fix_img, 1))
```

## 상하좌우 반전


```python
plt.imshow(cv2.flip(fix_img, -1))
```

## 그레이 영상 표시


```python
plt.imshow(img_gray, cmap='gray')
```

## 그레이 영상 저장


```python
cv2.imwrite('../data/lenna_gray.jpg', img_gray)
```
