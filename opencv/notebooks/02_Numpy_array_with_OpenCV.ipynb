## 라이브러리 준비


```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

## PIL 이미지 읽기


```python
pic = Image.open('../data/test.png')
pic
```

## PIL 타입


```python
type(pic)
```

## NumPy 배열 변환


```python
pic_arr = np.asarray(pic)
type(pic_arr)
```

## 배열 크기


```python
pic_arr.shape
```

## 전체 이미지 표시


```python
plt.imshow(pic_arr)
```

## R 채널과 viridis


```python
pic_red = pic_arr.copy()
plt.imshow(pic_red[:, :, 0])
```

## R 채널을 gray로 표시


```python
pic_red = pic_arr.copy()
plt.imshow(pic_red[:, :, 0], cmap='gray')
```

## G 채널


```python
pic_red = pic_arr.copy()
plt.imshow(pic_red[:, :, 1], cmap='gray')
```

## B 채널


```python
pic_red = pic_arr.copy()
plt.imshow(pic_red[:, :, 2], cmap='gray')
```

## G 채널 제거


```python
pic_red[:, :, 1] = 0
plt.imshow(pic_red)
```

## B 채널도 제거


```python
pic_red[:, :, 2] = 0
plt.imshow(pic_red)
```
