## 라이브러리 준비


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## 빈 영상 만들기


```python
blank_img = np.zeros((512,512,3),dtype=np.int8)
blank_img.shape
```

## 빈 영상 확인


```python
plt.imshow(blank_img)
```

## 사각형·원·선·문자


```python
cv2.rectangle(
    blank_img, pt1=(384,0),pt2 =(510,150), color=(0,255,0), thickness=10
)
cv2.circle(
    blank_img, center=(255,255), radius=50, color=(0,0,255), thickness=5
)

cv2.circle(
    blank_img, center=(100,100), radius=50, color=(255,0,255), thickness=-1
)

cv2.line(
    blank_img, pt1=(0,0), pt2=(512,512), color=(122,122,122), thickness=3
)
font = cv2.FONT_HERSHEY_TRIPLEX
cv2.putText(
    blank_img, 
    text="Im hungry", 
    org=(10,500), 
    fontFace=font, 
    fontScale=1, 
    color=(255,255,255), 
    thickness=2,
    lineType=cv2.LINE_AA

)


plt.imshow(blank_img)
```

## 다각형 꼭짓점


```python
blank_img = np.zeros((512,512,3),dtype=np.int32)
vertices = np.array([[100,300],[200,200],[400,300],[200,400]],dtype=np.int32)
vertices
```

## 꼭짓점 배열 크기


```python
vertices.shape
```

## polylines 입력 형태


```python
pts = vertices.reshape((-1,1,2))
pts.shape
```

## 원본 꼭짓점 확인


```python
vertices
```

## 변환된 꼭짓점 확인


```python
pts
```

## 다각형 그리기


```python
cv2.polylines(
    blank_img,
    [pts],
    isClosed=True,
    color=(0,255,0),
    thickness=5
)
plt.imshow(blank_img)
```
