> `cv2.imshow()`와 카메라·마우스 예제는 데스크톱 Jupyter/VS Code 환경에서 실행하는 것을 권장합니다.


## 카메라 장치 확인

Linux 터미널에서는 `ls /dev/video*`로 카메라 장치를 확인할 수 있습니다.


```python
import cv2

cap = cv2.VideoCapture(0)
if not cap.isOpened():
    raise RuntimeError("카메라를 열 수 없습니다.")

print("width:", int(cap.get(cv2.CAP_PROP_FRAME_WIDTH)))
print("height:", int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT)))
print("fps:", cap.get(cv2.CAP_PROP_FPS))

while True:
    ok, frame = cap.read()
    if not ok:
        break
    cv2.imshow("camera", frame)
    if cv2.waitKey(1) & 0xFF in (27, ord("q")):
        break

cap.release()
cv2.destroyAllWindows()
```

## 카메라 영상 저장

`q` 또는 ESC를 누르면 저장을 마칩니다.


```python
cap = cv2.VideoCapture(0)
if not cap.isOpened():
    raise RuntimeError("카메라를 열 수 없습니다.")

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = cap.get(cv2.CAP_PROP_FPS) or 30.0
fourcc = cv2.VideoWriter_fourcc(*"mp4v")
writer = cv2.VideoWriter("output.mp4", fourcc, fps, (width, height))

while True:
    ok, frame = cap.read()
    if not ok:
        break
    writer.write(frame)
    cv2.imshow("recording", frame)
    if cv2.waitKey(1) & 0xFF in (27, ord("q")):
        break

writer.release()
cap.release()
cv2.destroyAllWindows()
print("output.mp4 저장 완료")
```
